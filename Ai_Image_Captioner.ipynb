```python
import torch
from PIL import Image
from transformers import BlipProcessor, BlipForConditionalGeneration

# 1. Load the model and processor
# We use the 'base' model for speed; use 'large' for better accuracy
device = "cuda" if torch.cuda.is_available() else "cpu"
model_id = "Salesforce/blip-image-captioning-base"

processor = BlipProcessor.from_pretrained(model_id)
model = BlipForConditionalGeneration.from_pretrained(model_id).to(device)

def generate_caption(image_path):
    # 2. Load and preprocess the image

    raw_image = Image.open(image_path).convert('RGB')

    # 3. Process the image for the model
    # Conditional captioning (starts with a prompt)
    text = "a photography of"
    inputs = processor(raw_image, text, return_tensors="pt").to(device)

    # 4. Generate the caption
    out = model.generate(**inputs)

    # 5. Decode and print results
    caption = processor.decode(out[0], skip_special_tokens=True)
    return caption

```

```python
for filename in uploaded.keys():
    caption = generate_caption(filename)
    print(f"\nImage: {filename}")
    print(f"AI Caption: {caption}")
```
